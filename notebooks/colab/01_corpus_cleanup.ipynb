Upload `corpus.jsonl` + flagged rows from `reports/quality.csv`, or mount OneDrive.

```
# !pip install pymupdf openai python-dotenv
import json, os
from pathlib import Path

CORPUS_JSONL = Path('corpus.jsonl')  # upload from data/pyq-corpus/
OUT = Path('pyq-corpus-v1.jsonl')
OPENAI_API_KEY = os.environ.get('OPENAI_API_KEY')

def load_flagged(path):
    rows = []
    for line in path.read_text(encoding='utf-8').splitlines():
        r = json.loads(line)
        flags = r.get('flags') or []
        if flags:
            rows.append(r)
    return rows

# TODO: for each row, open pdf_path from papers table, clip page image, call vision API
# Write repaired records to OUT
print('Flagged count:', len(load_flagged(CORPUS_JSONL)) if CORPUS_JSONL.exists() else 'upload corpus.jsonl')
```
